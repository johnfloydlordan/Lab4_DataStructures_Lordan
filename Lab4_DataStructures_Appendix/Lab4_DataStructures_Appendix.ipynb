```python
# Exercise 1

from collections import defaultdict, Counter

SEED_NUM = 2
STUDENT_MAJOR = "[BSECE]"
FAVORITE_ARTIST = "[The Ridleys]"
LOCAL_HAZARD = "[TYPHOON]"
CONTROL_NUM = max(3, SEED_NUM)

stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE"),
]

# Add exactly CONTROL_NUM + 3 custom tuples
custom_applications = [
    ("TUP Manila", STUDENT_MAJOR),
    ("TUP Manila", FAVORITE_ARTIST.replace(" ", "_")),
    ("TUP Taguig", "BSCpE"),
    ("TUP Visayas", "BSIT"),
    ("TUP Manila", "BSECE"),
    ("TUP Manila", "BSME"),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", STUDENT_MAJOR),
    ("TUP Visayas", "BSEE"),
    ("TUP Manila", "BSECE"),
]

stream.extend(custom_applications[:CONTROL_NUM + 3])

# PHASE 1: AGGREGATION
campus_data = defaultdict(list)

for campus, program in stream:
    campus_data[campus].append(program)

# PHASE 2: FREQUENCY ANALYSIS
manila_counter = Counter(campus_data["TUP Manila"])
top_program, top_freq = manila_counter.most_common(1)[0]


# ASSESSMENT OUTPUT
print("Assessment Data:")
print("Field 1 - CONTROL_NUM Used:", CONTROL_NUM)
print("Field 2 - STUDENT_MAJOR Used:", STUDENT_MAJOR)
print("Field 3 - Expanded Stream Length (After adding data):", len(stream))
print("Field 4 - Total Applications Processed for TUP Manila:", len(campus_data["TUP Manila"]))
print("Field 5 - Top Requested Program at TUP Manila:", top_program)
print("Field 6 - Exact Frequency of Top Program:", top_freq)

print("\nGrouped Applications:")
for campus, programs in campus_data.items():
    print(f"{campus}: {programs}")

print("\nTUP Manila Frequency Counter:")
print(manila_counter)
```

```python
# Exercise 2

# OFFICIAL FESTIVAL LINEUP
festival = {
    "Ben&Ben",
    "SB19",
    "BINI",
    "Eraserheads",
    FAVORITE_ARTIST,
    "Zild",
    f"Indie Artist {CONTROL_NUM}"
}

# TWO USER PROFILES
user_a = {
    "Ben&Ben",
    "BINI",
    "Maki",
    "Dionela",
    FAVORITE_ARTIST
}

user_b = {
    "SB19",
    "Eraserheads",
    "Zild",
    f"Indie Artist {CONTROL_NUM}",
    "Parokya ni Edgar"
}

# Add exactly CONTROL_NUM new artists to the festival
new_artists = {
    "Cup of Joe",
    "Alamat",
    "Moira",
    "Arthur Nery",
    "Lola Amour",
    "December Avenue",
    "Rico Blanco",
    "IV of Spades"
}

festival.update(list(new_artists)[:CONTROL_NUM])


# JACCARD SIMILARITY
intersection_a = user_a.intersection(festival)
union_a = user_a.union(festival)
jaccard_a = (len(intersection_a) / len(union_a)) * 100

intersection_b = user_b.intersection(festival)
union_b = user_b.union(festival)
jaccard_b = (len(intersection_b) / len(union_b)) * 100

# Dealbreakers = artists User A listens to that are NOT in the updated festival
dealbreakers_a = user_a.difference(festival)


# ASSESSMENT OUTPUT
print("Assessment Data:")
print("Field 1 - Total Artists in Expanded Festival:", len(festival))
print("Field 2 - Length of Intersection (User A & Festival):", len(intersection_a))
print("Field 3 - Length of Union (User A & Festival):", len(union_a))
print("Field 4 - User A Jaccard Similarity (%):", round(jaccard_a, 2))
print("Field 5 - User B Jaccard Similarity (%):", round(jaccard_b, 2))
print("Field 6 - User A Dealbreaker Artists (Set Output):", dealbreakers_a)

print("\nExpanded Festival Lineup:")
print(festival)

print("\nUser A Intersection:")
print(intersection_a)

print("\nUser B Intersection:")
print(intersection_b)
```

```python
# Exercise 3

from collections import deque, Counter

# INITIALIZE DEQUE
alert_window = deque(maxlen=5)

# =========================
# BASE BURST OF ALERTS
# =========================
burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC"),
]

# Add exactly CONTROL_NUM + 2 custom alerts
next_timestamp = CONTROL_NUM + 7
custom_alerts = [
    (next_timestamp + 0, "FIRE"),
    (next_timestamp + 1, "WEATHER"),
    (next_timestamp + 2, "EARTHQUAKE"),
    (next_timestamp + 3, LOCAL_HAZARD),
    (next_timestamp + 4, "TRAFFIC"),
    (next_timestamp + 5, "WEATHER"),
    (next_timestamp + 6, "FIRE"),
    (next_timestamp + 7, "TRAFFIC"),
    (next_timestamp + 8, "WEATHER"),
    (next_timestamp + 9, LOCAL_HAZARD),
]

burst.extend(custom_alerts[:CONTROL_NUM + 2])

# PROCESS ALERTS WITH SPAM RULE
# Reject if category already appears 2 or more times in current deque
weather_plus3_appended = "N"
local_hazard_appended = "N"

for timestamp, category in burst:
    existing_categories = [item[1] for item in alert_window]
    category_counter = Counter(existing_categories)

    if category_counter[category] >= 2:
        # reject
        continue
    else:
        alert_window.append((timestamp, category))

        if timestamp == CONTROL_NUM + 3 and category == "WEATHER":
            weather_plus3_appended = "Y"

        if category == LOCAL_HAZARD:
            local_hazard_appended = "Y"

# FINAL SUM OF TIMESTAMPS
timestamp_sum = sum(item[0] for item in alert_window)


# ASSESSMENT OUTPUT
print("Assessment Data:")
print("Field 1 - LOCAL_HAZARD Used:", LOCAL_HAZARD)
print("Field 2 - Total Burst Size (After adding custom alerts):", len(burst))
print(f'Field 3 - Did "WEATHER" at Timestamp {CONTROL_NUM + 3} append? (Y/N):', weather_plus3_appended)
print("Field 4 - Did LOCAL_HAZARD append? (Y/N):", local_hazard_appended)
print("Field 5 - Final Deque Length:", len(alert_window))
print("Field 6 - Final Sum of Timestamps in Deque:", timestamp_sum)
print("Field 7 - Exact Final Deque Output:", list(alert_window))

print("\nExecution Log:")
for item in alert_window:
    print(item)
```
